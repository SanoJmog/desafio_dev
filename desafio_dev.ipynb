```python
import pandas as pd

def calculo_comissao(valor):
  valor0 = 0
  valor1 = 0
  valor2 = 0
  if (valor > 500):
    valor2 = valor * 0.05
  elif (valor > 100):
      valor1 = valor * 0.01
  else:
      return 0.0
  total = valor1 + valor2
  return total
df = pd.read_csv('vendas_limpo.csv')

df['comissao'] = df['valor'].apply(calculo_comissao)

comissao_total = df.groupby('vendedor')['comissao'].sum().reset_index()

comissao_total.index = comissao_total.index + 1
comissao_total['comissao'] = comissao_total['comissao'].apply(
    lambda x: f"R$ {x:,.2f}".replace('.', ','))
print(comissao_total);
```
